```python
from sklearn.datasets import make_regression
```

```python
x,y = make_regression(n_samples=200, n_features=1, noise=30)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x,y)
plt.show
```

Após a execução seguinte, o objeto modelo que acabou de ser criado acima está pronto para receber os dados que darão origem ao modelo. Como não foi indicado nenhum parâmetro específico na função, estamos utilizando suas configurações padrão.

```python
from sklearn.linear_model import LinearRegression
#Criação do modelo
modelo = LinearRegression()
```

```python
modelo.fit(x, y)
```

```python
modelo.predict(x)
```

```python
plt.scatter(x,y)
plt.plot(x, modelo.predict(x), color='red', linewidth=3)
plt.show()
```

Tudo o que está próximo a reta melhor representa o conjunto de amostra, o que está mais longe não apresenta boas características para o conjunto
